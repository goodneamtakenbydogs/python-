### 运行并观察

运行本题代码，不覆盖原来的预测。

```python
from functools import lru_cache, partial
def render(case_id, *, prefix):
    return prefix + case_id
format_case = partial(render, prefix="case:")
calls = []
@lru_cache(maxsize=2)
def parse_count(text):
    calls.append(text)
    return int(text)
parse_count("3")
parse_count("3")
before_clear = len(calls)
parse_count.cache_clear()
parse_count("3")
actual_25 = (format_case("A"), before_clear, len(calls))
print(*actual_25, sep="\n")
```
