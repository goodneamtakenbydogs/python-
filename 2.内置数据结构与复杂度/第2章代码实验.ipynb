## 实验 6：哪些常见对象可哈希

运行前逐项判断。这里检查的是对象能否稳定参与 `hash()`，不是简单背诵“可变或不可变”四个字。最后再观察类型不同但数值相等的 `1` 和 `1.0` 为什么必须具有相同哈希值；`__eq__()` 与 `__hash__()` 的完整自定义规则在第 6 章学习。

```python
def is_hashable(value):
    try:
        hash(value)
    except TypeError:
        return False
    return True

hashable_results = {
    "integer": is_hashable(10),
    "text": is_hashable("key"),
    "plain_tuple": is_hashable((1, 2)),
    "list": is_hashable([1, 2]),
    "dictionary": is_hashable({"a": 1}),
    "set": is_hashable({1, 2}),
    "frozen_set": is_hashable(frozenset({1, 2})),
}
print(hashable_results)
print(1 == 1.0)
print(hash(1) == hash(1.0))

observation_06 = {
    "01. 整数是否可哈希": "__未填写__",    "02. 字符串是否可哈希": "__未填写__",    "03. 普通元组是否可哈希": "__未填写__",    "04. 列表是否可哈希": "__未填写__",    "05. 字典是否可哈希": "__未填写__",    "06. 集合是否可哈希": "__未填写__",    "07. frozenset 是否可哈希": "__未填写__",    "08. 1 == 1.0 的结果": "__未填写__",    "09. hash(1) == hash(1.0) 的结果": "__未填写__",}
```

## 实验 15：大小为 `k` 的 Top K 堆

```python
values = [4, 5, 8, 2]
k = 3
top_k_heap = []

for value in values:
    heapq.heappush(top_k_heap, value)
    if len(top_k_heap) > k:
        heapq.heappop(top_k_heap)

kept_after_initial = sorted(top_k_heap)
third_largest_initial = top_k_heap[0]

heapq.heappush(top_k_heap, 10)
if len(top_k_heap) > k:
    heapq.heappop(top_k_heap)

print(kept_after_initial, third_largest_initial)
print(sorted(top_k_heap), top_k_heap[0])

observation_15 = {
    "01. 处理初始 values 后保留的三个数": "__未填写__",    "02. 初始数据中的第 3 大": "__未填写__",    "03. 加入 10 后保留的三个数": "__未填写__",    "04. 加入 10 后的第 3 大": "__未填写__",}
```

## 可选实验：列表与集合成员查找的耗时对比

下面比较列表和集合成员查找的测量结果。实际秒数会受到机器和运行状态影响，因此不要填写固定答案。尝试改变 `size`，观察增长趋势，并解释为什么一次测量不能证明严格复杂度。

```python
from timeit import timeit

for size in [1_000, 10_000, 100_000]:
    list_values = list(range(size))
    set_values = set(list_values)
    missing = -1
    list_time = timeit(lambda: missing in list_values, number=200)
    set_time = timeit(lambda: missing in set_values, number=200)
    print(size, list_time, set_time)
```
