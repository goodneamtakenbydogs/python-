## 实验 10：包初始化与子模块绑定

临时创建普通包，观察 `__init__.py` 与子模块的执行顺序及父包属性。

```python
import importlib
import sys
import tempfile
from pathlib import Path

package_name = "chapter8_pkg_observation"
with tempfile.TemporaryDirectory() as directory:
    root = Path(directory)
    package = root / package_name
    package.mkdir()
    (package / "__init__.py").write_text('EVENTS = ["package"]\n', encoding="utf-8")
    (package / "child.py").write_text('from . import EVENTS\nEVENTS.append("submodule")\nANSWER = 42\n', encoding="utf-8")
    sys.path.insert(0, directory)
    try:
        child = importlib.import_module(package_name + ".child")
        parent = sys.modules[package_name]
        events_result = parent.EVENTS.copy()
        answer = child.ANSWER
        parent_has_child = parent.child is child
    finally:
        sys.path.remove(directory)
        for name in list(sys.modules):
            if name == package_name or name.startswith(package_name + "."):
                del sys.modules[name]
print(events_result, answer, parent_has_child)

observation_10 = {
    "01. 包与子模块的事件顺序": "__未填写__",
    "02. 子模块 ANSWER 的值": "__未填写__",
    "03. 父包 child 属性是否指向子模块": "__未填写__",
}
```

## 实验 13：循环导入与部分初始化

临时包让两个模块在顶层互相依赖，并让后加载者立即读取尚未建立的属性。

```python
import importlib
import sys
import tempfile
from pathlib import Path

package_name = "chapter8_cycle_pkg"
with tempfile.TemporaryDirectory() as directory:
    package = Path(directory, package_name)
    package.mkdir()
    (package / "__init__.py").write_text("", encoding="utf-8")
    (package / "a.py").write_text("from . import b\nREADY = True\n", encoding="utf-8")
    (package / "b.py").write_text("from . import a\nSEEN = a.READY\n", encoding="utf-8")
    sys.path.insert(0, directory)
    try:
        try:
            importlib.import_module(package_name + ".a")
        except Exception as error:
            error_name = type(error).__name__
            mentions_partial = "partially initialized module" in str(error)
    finally:
        sys.path.remove(directory)
        for name in list(sys.modules):
            if name == package_name or name.startswith(package_name + "."):
                del sys.modules[name]
print(error_name, mentions_partial)

observation_13 = {
    "01. 循环导入记录的异常名称": "__未填写__",
    "02. 错误消息是否提到 partially initialized module": "__未填写__",
}
```
