## 补充实验 16：`not`、运算优先级与求值顺序

运行前先写出每个值，并分别说明：表达式怎样按优先级分组、函数按什么顺序调用、哪个调用因为短路而没有发生。`not in` 和 `is not` 应作为完整比较运算符阅读。

```python
not_empty_text = not ""
not_nonempty_list = not [0]
arithmetic_precedence = 2 + 3 * 4
arithmetic_parenthesized = (2 + 3) * 4
logic_precedence = True or False and False
comparison_before_not = not 1 == 2
membership_not_in = "java" not in "python"
identity_is_not = None is not None

evaluation_order = []

def mark(label, value):
    evaluation_order.append(label)
    return value

evaluation_result = (
    mark("first", 1)
    + mark("second", 2) * mark("third", 3)
)

short_circuit_order = []

def mark_short(label, value):
    short_circuit_order.append(label)
    return value

short_circuit_result = (
    mark_short("left", True)
    or mark_short("right", False)
)

print(not_empty_text, not_nonempty_list)
print(arithmetic_precedence, arithmetic_parenthesized)
print(logic_precedence, comparison_before_not)
print(membership_not_in, identity_is_not)
print(evaluation_order, evaluation_result)
print(short_circuit_order, short_circuit_result)

observation_16 = {
    "01. not '' 的结果": "__未填写__",    "02. not [0] 的结果": "__未填写__",    "03. 2 + 3 * 4 的结果": "__未填写__",    "04. (2 + 3) * 4 的结果": "__未填写__",    "05. True or False and False 的结果": "__未填写__",    "06. not 1 == 2 的结果": "__未填写__",    "07. 'java' not in 'python' 的结果": "__未填写__",    "08. None is not None 的结果": "__未填写__",    "09. mark 的调用标签顺序": "__未填写__",    "10. 三个 mark 表达式的计算结果": "__未填写__",    "11. 短路表达式中的调用标签顺序": "__未填写__",    "12. 短路表达式的计算结果": "__未填写__",}
```
