## 实验 13：`global` 与 `nonlocal`

```python
global_count = 10

def change_global():
    global global_count
    global_count += 1

def run_nonlocal_counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    return increase(), increase()

change_global()
nonlocal_results = run_nonlocal_counter()
print(global_count, nonlocal_results)

observation_13 = {
    "01. change_global() 后的 global_count": "__未填写__",    "02. 连续两次调用 increase() 的结果": "__未填写__",}
```

## 实验 15：函数对象、高阶函数与注解

```python
def add(left: int, right: int) -> int:
    return left + right

def apply_twice(operation, value):
    return operation(operation(value))

def add_one(number):
    return number + 1

stored_function = add
stored_call = stored_function(3, 5)
higher_order_result = apply_twice(add_one, 3)
annotation_keys = sorted(add.__annotations__.keys())
annotated_string_call = add("a", "b")
print(stored_call, higher_order_result)
print(annotation_keys, annotated_string_call)

observation_15 = {
    "01. stored_function(3, 5) 的结果": "__未填写__",    "02. apply_twice(add_one, 3) 的结果": "__未填写__",    "03. add.__annotations__ 的键": "__未填写__",    "04. add('a', 'b') 的结果": "__未填写__",}
```

## 实验 16：递归调用与逐层返回

运行前写出完整 `trace`。区分进入调用的顺序与返回调用的顺序，并说明每层栈帧在等待什么。

```python
recursion_trace = []

def factorial(number):
    recursion_trace.append(("enter", number))
    if number == 0:
        recursion_trace.append(("return", 0, 1))
        return 1

    result = number * factorial(number - 1)
    recursion_trace.append(("return", number, result))
    return result

factorial_result = factorial(3)
print(factorial_result)
for event in recursion_trace:
    print(event)

observation_16 = {
    "01. factorial(3) 的结果": "__未填写__",    "02. recursion_trace 的完整内容": "__未填写__",}
```
