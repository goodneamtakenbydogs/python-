第一种只有标注，第二种显式拒绝布尔数量。

### 运行前预测

先读源码，不运行时填写。

```python
prediction_02 = {
    "01. Plain.count 的类型名称": "__未填写__",
    "02. 严格构造捕获的异常名称": "__未填写__",
}
```

### 运行并观察

保留原预测，运行本题源码。

```python
from dataclasses import dataclass


@dataclass
class Plain:
    count: int

@dataclass
class Strict:
    count: int
    def __post_init__(self):
        if type(self.count) is not int:
            raise ValueError("exact integer required")

plain = Plain(True)
try:
    Strict(True)
except ValueError as error:
    error_name = type(error).__name__
actual_02 = (type(plain.count).__name__, error_name)
print(*actual_02, sep='\n')
```

## 实验 14：记录上限与停止消费

输入生成器有十条，配置上限为二；记录实际被请求的项。

### 运行前预测

先读源码，不运行时填写。

```python
prediction_14 = {
    "01. 异常名称": "__未填写__",
    "02. 实际消费数量": "__未填写__",
}
```

### 运行并观察

保留原预测，运行本题源码。

```python
from contextlib import closing

from chapter16_checks import load_example

model = load_example("models")
engine = load_example("engine")
config_module = load_example("config")
errors = load_example("errors")
consumed = []
def source():
    for index in range(10):
        consumed.append(index)
        yield model.CaseRecord(f"C-{index}", "skipped", 0)
with closing(source()) as records:
    try:
        engine.summarize(records, config_module.Config(max_records=2))
    except errors.InputError as error:
        error_name = type(error).__name__
actual_14 = (error_name, len(consumed))
print(*actual_14, sep='\n')
```
