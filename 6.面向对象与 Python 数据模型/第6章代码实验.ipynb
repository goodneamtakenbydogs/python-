## 实验 13：相等、哈希与可变键风险

先区分身份和相等，再观察不支持的比较与哈希状态。最后只记录键的哈希变化，不依赖一次字典查找是否碰巧成功。

```python
class Label:
    def __init__(self, text):
        self.text = text

    def __eq__(self, other):
        if not isinstance(other, Label):
            return NotImplemented
        return self.text == other.text

first = Label("A")
second = Label("A")
error_name = None
try:
    hash(first)
except Exception as error:
    error_name = type(error).__name__
print(first == second, first is second, first == object())
print(Label.__hash__ is None, error_name)

class MutableKey:
    def __init__(self, code):
        self.code = code

    def __eq__(self, other):
        if not isinstance(other, MutableKey):
            return NotImplemented
        return self.code == other.code

    def __hash__(self):
        return hash(self.code)

key = MutableKey(1)
lookup = {key: "saved"}
before = hash(key)
key.code = 2
after = hash(key)
hash_changed = before != after
print(hash_changed, len(lookup))

observation_13 = {
    "01. first == second 的结果": "__未填写__",
    "02. first is second 的结果": "__未填写__",
    "03. first == object() 的结果": "__未填写__",
    "04. Label.__hash__ is None 的结果": "__未填写__",
    "05. hash(first) 记录的异常名称": "__未填写__",
    "06. 修改前后哈希是否不同": "__未填写__",
    "07. 修改后 len(lookup) 的结果": "__未填写__",
}
```
