## 实验 7：生成器保存局部状态

逐次恢复同一个生成器，观察局部名字 `current` 是否从头建立，还是从暂停位置继续。

```python
def count_up(limit):
    current = 1
    while current <= limit:
        yield current
        current += 1

generator = count_up(3)
first_value = next(generator)
second_value = next(generator)
remaining = list(generator)
after_end = next(generator, "结束")

print(first_value, second_value, remaining, after_end)

observation_07 = {
    "01. 第一次 next(generator) 的结果": "__未填写__",
    "02. 第二次 next(generator) 的结果": "__未填写__",
    "03. 随后 list(generator) 的结果": "__未填写__",
    "04. 耗尽后 next(generator, '结束') 的结果": "__未填写__",
}
```
