# 第 14 章代码实验：Python 测试框架与用例组织

先读讲义与示例工程。观察题使用 CPython 3.12.13、pytest 9.1.1；除实验 19 直接使用标准库外，均在独立子进程的临时工程中实际运行 pytest，不在 Notebook 内核反复调用 pytest.main。

run_pytest_project 是本章辅助接口，不是 pytest 自带 API。输入文件路径、源码和命令实参；返回 exit_code、nodeids、counts、output 与指定 artifacts 文本。nodeids 是筛选后测试项；counts 按收集或执行阶段报告累加，清理错误可能与测试体 passed 同时出现。运行器关闭第三方插件自动加载、排除外部 pytest 选项；示例只操作临时目录。它不是任意代码的系统沙箱。

先读源码与运行实参 → 填 prediction → 运行 → 填 observation → 本题即时检查 → 写原因。把整个 "__未填写__"（包括引号）换成实际 Python 值，保留两位字段编号。错误预测保留原样；检查器只要求预测已填写，不会要求改成正确答案。判断题结果形态也由自己预测。

观察代码会打印本题要记录的 actual_NN 各字段；需要完整报错时查看 result_NN["output"]。部分错误是故意安排的，子进程非零退出是观察目标，不代表内核应中断。实现题返回源码文本，检查器会建立临时模块并实际执行；运行需数秒至数十秒。

共 24 道观察题、4 道实现题，每题有即时检查。最后人工复盘不计入 run_checks 的通过结论。公开笔记只使用虚构数据。
